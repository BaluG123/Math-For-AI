```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)
```

```python
# Ex 1
d1 = np.random.randint(1,7,10000)
d2 = np.random.randint(1,7,10000)
print(f"P(sum=7) ≈ {np.mean(d1+d2==7):.4f} (exact: {6/36:.4f})")
```

```python
# Ex 2
print(f"P(A∪B) = 0.4 + 0.5 - 0.1 = {0.4+0.5-0.1}")
```

```python
# Ex 3
s = np.random.uniform(0, 10, 50000)
print(f"Mean: {s.mean():.4f} (theory: 5)")
print(f"Var: {s.var():.4f} (theory: {100/12:.4f})")
print(f"Std: {s.std():.4f} (theory: {np.sqrt(100/12):.4f})")
```

```python
# Ex 4
X = np.random.randn(100000)
Y = np.random.exponential(2, 100000)
a, b = 3, -2
print(f"E[{a}X + {b}Y] = {np.mean(a*X + b*Y):.4f}")
print(f"{a}E[X] + {b}E[Y] = {a*np.mean(X) + b*np.mean(Y):.4f}")
```

```python
# Ex 5
k = np.arange(0, 15)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.bar(k, stats.poisson.pmf(k, 5), color='steelblue'); ax1.set_title('Poisson PMF')
ax2.step(k, stats.poisson.cdf(k, 5), color='red'); ax2.set_title('Poisson CDF')
plt.tight_layout(); plt.show()
```

```python
# Ex 6: Bayes spam filter
P_spam = 0.3
P_free_spam = 0.8
P_free_not = 0.1
P_free = P_free_spam*P_spam + P_free_not*(1-P_spam)
P_spam_free = P_free_spam*P_spam / P_free
print(f"P(spam|'free') = {P_spam_free:.4f}")
```

```python
# Ex 7: CLT with Exponential
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for idx, n in enumerate([1, 5, 30, 100]):
    means = np.mean(np.random.exponential(0.5, (10000, n)), axis=1)
    axes[idx].hist(means, bins=50, density=True, alpha=0.7)
    axes[idx].set_title(f'n={n}')
plt.suptitle('CLT with Exponential'); plt.tight_layout(); plt.show()
```

```python
# Ex 8
X = np.random.randn(1000)
Y = 2*X + np.random.randn(1000)
Z = -X + 0.5*np.random.randn(1000)
data = np.column_stack([X, Y, Z])
print(f"Cov matrix:\n{np.round(np.cov(data.T), 3)}")
print(f"Corr matrix:\n{np.round(np.corrcoef(data.T), 3)}")
```

```python
# Ex 9: Total probability
P_def = 0.5*0.02 + 0.3*0.03 + 0.2*0.05
P_C_def = (0.05*0.2) / P_def
print(f"P(defective) = {P_def}")
print(f"P(from C | defective) = {P_C_def:.4f}")
```

```python
# Ex 10
X = np.random.randn(100000)
Y_indep = np.random.randn(100000)
Y_dep = 2*X  # Perfectly dependent
print(f"Independent: Var(X+Y)={np.var(X+Y_indep):.3f}, Var(X)+Var(Y)={np.var(X)+np.var(Y_indep):.3f}")
print(f"Dependent: Var(X+Y)={np.var(X+Y_dep):.3f}, Var(X)+Var(Y)={np.var(X)+np.var(Y_dep):.3f}")
```

```python
# Ex 11: Bayesian coin flip → Beta posterior
heads, total = 7, 10
theta = np.linspace(0, 1, 200)
prior = stats.beta.pdf(theta, 1, 1)  # Uniform = Beta(1,1)
posterior = stats.beta.pdf(theta, 1+heads, 1+total-heads)

plt.plot(theta, prior, 'b--', label='Prior Beta(1,1)')
plt.plot(theta, posterior, 'r-', lw=2, label=f'Posterior Beta({1+heads},{1+total-heads})')
plt.axvline(heads/total, color='g', ls=':', label=f'MLE = {heads/total}')
plt.legend(); plt.title('Bayesian Coin Flip'); plt.xlabel('θ'); plt.show()
```

```python
# Ex 12: Monte Carlo π
N = 100000
x = np.random.uniform(0, 1, N)
y = np.random.uniform(0, 1, N)
inside = x**2 + y**2 <= 1
pi_est = 4 * np.cumsum(inside) / np.arange(1, N+1)
plt.plot(pi_est); plt.axhline(np.pi, color='r', ls='--')
plt.xscale('log'); plt.title(f'π ≈ {pi_est[-1]:.5f}'); plt.show()
```

```python
# Ex 13: Simple Naive Bayes
spam_words = {'free': 8, 'win': 6, 'hello': 2, 'meeting': 1}
ham_words = {'free': 1, 'win': 1, 'hello': 8, 'meeting': 7}
total_spam = sum(spam_words.values())
total_ham = sum(ham_words.values())

new_msg = ['free', 'win']
P_spam = 0.4
log_p_spam = np.log(P_spam)
log_p_ham = np.log(1 - P_spam)
vocab = set(list(spam_words.keys()) + list(ham_words.keys()))
for word in new_msg:
    log_p_spam += np.log((spam_words.get(word, 0) + 1) / (total_spam + len(vocab)))
    log_p_ham += np.log((ham_words.get(word, 0) + 1) / (total_ham + len(vocab)))

print(f"log P(spam|msg) ∝ {log_p_spam:.4f}")
print(f"log P(ham|msg)  ∝ {log_p_ham:.4f}")
print(f"Prediction: {'SPAM' if log_p_spam > log_p_ham else 'HAM'}")
```
