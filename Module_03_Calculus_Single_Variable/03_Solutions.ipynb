# Module 3 Solutions: Calculus — Single Variable

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy import integrate
x = sp.Symbol('x')
```

```python
# Ex 1: Limits
print(sp.limit(sp.sin(3*x)/x, x, 0))         # 3
print(sp.limit((x**2+1)/(2*x**2-3), x, sp.oo)) # 1/2
print(sp.limit((sp.exp(x)-1-x)/x**2, x, 0))    # 1/2
```

```python
# Ex 2: Derivatives
f = 3*x**4 - 2*x**2 + 7
g = sp.sin(x) * sp.exp(x)
h = x**2 / (x + 1)
for func, name in [(f,'f'), (g,'g'), (h,'h')]:
    d = sp.diff(func, x)
    print(f"{name}'(x) = {d}, at x=2: {float(d.subs(x, 2)):.4f}")
```

```python
# Ex 3: Numerical differentiation comparison
f = lambda t: t**3
exact = 12.0  # 3*2^2
print(f"{'h':>10} {'Forward':>15} {'Central':>15}")
for h in [0.1, 0.01, 0.001, 0.0001]:
    fwd = (f(2+h) - f(2)) / h
    cen = (f(2+h) - f(2-h)) / (2*h)
    print(f"{h:>10} {abs(fwd-exact):>15.10f} {abs(cen-exact):>15.10f}")
```

```python
# Ex 4: Chain rule
funcs = [sp.sin(x**2+1), sp.exp(sp.cos(x)), sp.log(sp.sin(x))]
for func in funcs:
    print(f"d/dx [{func}] = {sp.diff(func, x)}")
```

```python
# Ex 5: Integrals
print(sp.integrate(x**3, (x, 0, 1)))          # 1/4
print(sp.integrate(sp.sin(x), (x, 0, sp.pi))) # 2
print(sp.integrate(1/x, (x, 1, sp.E)))        # 1
```

```python
# Ex 6: Taylor series of e^x
taylor5 = sp.series(sp.exp(x), x, 0, 6)
print(f"Taylor(5): {taylor5}")

x_vals = np.linspace(-3, 3, 200)
plt.figure(figsize=(10,6))
plt.plot(x_vals, np.exp(x_vals), 'k-', lw=3, label='e^x')
for n in [1, 2, 3, 5]:
    taylor = sum(x_vals**k / np.math.factorial(k) for k in range(n+1))
    plt.plot(x_vals, taylor, '--', lw=1.5, label=f'Order {n}')
plt.ylim(-2, 15); plt.legend(); plt.title('Taylor approx of e^x'); plt.grid(True); plt.show()
```

```python
# Ex 7: Critical points
f = x**3 - 6*x**2 + 9*x + 1
f1 = sp.diff(f, x)
f2 = sp.diff(f, x, 2)
cps = sp.solve(f1, x)
for cp in cps:
    nature = 'min' if f2.subs(x, cp) > 0 else 'max'
    print(f"x={cp}: f''={f2.subs(x,cp)} → {nature}, f({cp})={f.subs(x,cp)}")
```

```python
# Ex 8: Sigmoid properties
sigma = 1 / (1 + sp.exp(-x))
sigma_prime = sp.diff(sigma, x)
print(f"σ'(x) = {sp.simplify(sigma_prime)}")
print(f"σ(x)(1-σ(x)) = {sp.simplify(sigma * (1-sigma))}")
print(f"Equal? {sp.simplify(sigma_prime - sigma*(1-sigma)) == 0}")
# σ(-x) = 1 - σ(x)
print(f"σ(-x) = {sp.simplify(sigma.subs(x, -x))}")
print(f"1-σ(x) = {sp.simplify(1 - sigma)}")
```

```python
# Ex 9: Integration
print(sp.integrate(x * sp.exp(x**2), x))  # e^(x²)/2
print(sp.integrate(sp.cos(x)/(1+sp.sin(x)), x))  # log(1+sin(x))
print(sp.integrate(x * sp.exp(-x), (x, 0, sp.oo)))  # 1
```

```python
# Ex 10: Learning rate comparison
f = lambda t: t**4 - 3*t**3 + 2
fp = lambda t: 4*t**3 - 9*t**2

plt.figure(figsize=(10,6))
for lr in [0.001, 0.01, 0.05]:
    t = 4.0
    losses = []
    for _ in range(200):
        losses.append(f(t))
        t -= lr * fp(t)
    plt.plot(losses, label=f'lr={lr}')
plt.xlabel('Step'); plt.ylabel('f(x)'); plt.legend(); plt.title('LR Comparison'); plt.grid(True); plt.show()
```

```python
# Ex 11: Newton's method
f = lambda t: (t-2)**4
fp = lambda t: 4*(t-2)**3
fpp = lambda t: 12*(t-2)**2

# Gradient descent
x_gd = 5.0
gd_loss = []
for _ in range(50):
    gd_loss.append(f(x_gd))
    x_gd -= 0.01 * fp(x_gd)

# Newton
x_n = 5.0
n_loss = []
for _ in range(50):
    n_loss.append(f(x_n))
    if fpp(x_n) != 0:
        x_n -= fp(x_n) / fpp(x_n)

plt.semilogy(gd_loss, label='GD'); plt.semilogy(n_loss, label='Newton')
plt.legend(); plt.title('Newton vs GD'); plt.grid(True); plt.show()
```

```python
# Ex 12: Softmax Jacobian
def softmax(z):
    e = np.exp(z - np.max(z))
    return e / e.sum()

z = np.array([2.0, 1.0, 0.1])
s = softmax(z)
print(f"softmax = {np.round(s, 4)}")

# Numerical Jacobian
J = np.zeros((3, 3))
h = 1e-7
for j in range(3):
    z_plus = z.copy(); z_plus[j] += h
    J[:, j] = (softmax(z_plus) - s) / h
print(f"Jacobian:\n{np.round(J, 4)}")

# Verify: diag = s_i(1-s_i), off-diag = -s_i*s_j
J_expected = np.diag(s) - np.outer(s, s)
print(f"Match? {np.allclose(J, J_expected, atol=1e-5)}")
```

```python
# Ex 13: Cross-entropy derivative
z_sym, y_sym = sp.symbols('z y')
sig = 1 / (1 + sp.exp(-z_sym))
L = -(y_sym * sp.log(sig) + (1 - y_sym) * sp.log(1 - sig))
dL_dz = sp.simplify(sp.diff(L, z_sym))
print(f"∂L/∂z = {dL_dz}")
# Should simplify to σ(z) - y
```

```python
# Ex 14: Numerical integration comparison
f_np = lambda t: np.exp(-t**2)
exact = float(sp.integrate(sp.exp(-x**2), (x, 0, 1)))

# Trapezoidal
t = np.linspace(0, 1, 101)
trap = np.trapz(f_np(t), t)

# Simpson
simp, _ = integrate.quad(f_np, 0, 1)

# Monte Carlo
np.random.seed(42)
samples = np.random.uniform(0, 1, 10000)
mc = np.mean(f_np(samples))

print(f"Exact:       {exact:.8f}")
print(f"Trapezoidal: {trap:.8f} (error: {abs(trap-exact):.2e})")
print(f"Simpson:     {simp:.8f} (error: {abs(simp-exact):.2e})")
print(f"Monte Carlo: {mc:.8f} (error: {abs(mc-exact):.2e})")
```

```python
# Ex 15: Linear regression with gradient descent
np.random.seed(42)
X = np.random.randn(50)
y = 2 * X + 1 + 0.3 * np.random.randn(50)

w, b = 0.0, 0.0
lr = 0.01
losses = []

for step in range(500):
    y_pred = w * X + b
    loss = np.mean((y - y_pred)**2)
    losses.append(loss)
    dw = -2 * np.mean(X * (y - y_pred))
    db = -2 * np.mean(y - y_pred)
    w -= lr * dw
    b -= lr * db

print(f"Learned: w={w:.4f}, b={b:.4f} (true: w=2, b=1)")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.scatter(X, y, s=20, alpha=0.5)
ax1.plot(np.sort(X), w*np.sort(X)+b, 'r-', lw=2)
ax1.set_title('Learned Line'); ax1.grid(True)
ax2.plot(losses); ax2.set_title('Loss Curve'); ax2.set_xlabel('Step'); ax2.grid(True)
plt.tight_layout(); plt.show()
```
